```python
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-5"
msgtable = []
system = "you are  patient math tutor, i want you to direct the student on how to solve math questions"

def addmsg(msg, text):
    user_msg = {"role": "user", "content": text}
    msg.append(user_msg)

def assitmsg(msg, text):
    assit_msg = {"role": "assistant", "content": text}  # ✅ fixed spelling
    msg.append(assit_msg)

def getmsg(msg):
    response = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=msg
    )
    return response.content[0].text

```
